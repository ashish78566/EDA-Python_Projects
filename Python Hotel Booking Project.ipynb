### 🧾 Problem Statement

Hotel booking cancellations create major challenges for hotel management, affecting revenue forecasting, inventory control, and operational efficiency. To address these challenges, it’s essential to analyze booking data and identify the factors that influence whether a reservation is likely to be canceled.

This project analyzes a dataset containing hotel booking information — including customer details, reservation status, stay duration, room types, and booking channels — to uncover patterns and relationships that impact booking cancellations in City Hotels and Resort Hotels.

### Data Analysis and Visualization

```python
cancelled_perc=df['is_canceled'].value_counts(normalize=True)  #hotel booking cancelled and noncancelled percentage show
print(cancelled_perc)
```

```python
df['is_canceled'].value_counts()  # 74745 booking not cancelled
                                  #44152  booking is cancelled  
```

```python
plt.figure(figsize=(4,4))
plt.title('Reservation status count')
plt.bar(['Not Cancelled','Cancelled'],df['is_canceled'].value_counts(),edgecolor='k',width=0.7)
plt.show()

```

```python
plt.figure(figsize=(5,3))
ax1=sns.countplot(x='hotel',hue='is_canceled',data=df,palette='Oranges')
plt.title('Reservation status count',size=20)
plt.xlabel('hotel')
plt.ylabel('No of reservation')                            #This code makes a count plot that compares canceled 
                                                           #vs not canceled reservations for each hotel type (e.g., City Hotel, Resort Hotel) using Seaborn.
plt.legend(['not canceled','canceled'])
plt.show()
```

```python
resort_hotel=df[df['hotel']=='Resort Hotel']
resort_hotel['is_canceled'].value_counts(normalize=True)  #This code filters only Resort Hotel bookings and shows the percentage of canceled 
                                                          #vs not canceled reservations.
```

```python
city_hotel=df[df['hotel']=='City Hotel']                      #This code filters only City Hotel bookings and shows the percentage of canceled vs not canceled reservations.
city_hotel['is_canceled'].value_counts(normalize=True)
```

```python
resort_hotel=resort_hotel.groupby('reservation_status_date')[['adr']].mean()     #Groups bookings by date and calculates the average ADR (room price) for Resort Hotel and City Hotel separately.
city_hotel=city_hotel.groupby('reservation_status_date')[['adr']].mean()
```

```python
plt.figure(figsize=(10,3))
plt.title('Average Daily Rate in City and Resort Hotel',fontsize=20)
plt.plot(resort_hotel.index,resort_hotel['adr'],label='Resort Hotel')        #adr=average daily adte
plt.plot(city_hotel.index,city_hotel['adr'],label='City Hotel')              #Plots a line chart showing how the average daily rate (ADR) changes over time for Resort Hotel and City Hotel.
plt.legend(fontsize=20)
plt.show()
```

```python
df['month']=df['reservation_status_date'].dt.month
plt.figure(figsize=(10,3))
ax1=sns.countplot(x='month',hue='is_canceled',data=df,palette='rainbow')
plt.title('Reservation status month',size=20)
plt.xlabel('month')                                         #This code creates a bar chart showing the number of reservations per month, split by canceled vs not canceled bookings.
plt.ylabel('No of reservation')
plt.legend(['not canceled','canceled'])
plt.show()
```

```python
monthly_adr = df[df['is_canceled'] == 1].groupby('month')[['adr']].sum().reset_index()

# Correct sns.barplot call
sns.barplot(x='month', y='adr', data=monthly_adr)

plt.title('ADR per month', size=15)
plt.show()
```

```python
cancelled_data=df[df['is_canceled']==1]
top_10_country=cancelled_data['country'].value_counts()[:10]
plt.figure(figsize=(6,6))
plt.title('Top 10 countries with reservation',size=20)
plt.pie(top_10_country,autopct='%.2f',labels=top_10_country.index)
plt.show()
```

```python
cancelled_df_adr=cancelled_data.groupby('reservation_status_date')[['adr']].mean()
cancelled_df_adr.reset_index(inplace=True)
cancelled_df_adr.sort_values('reservation_status_date',inplace=True)

not_cancelled_data=df[df['is_canceled']==0]

not_cancelled_df_adr=not_cancelled_data.groupby('reservation_status_date')[['adr']].mean()
not_cancelled_df_adr.reset_index(inplace=True)
not_cancelled_df_adr.sort_values('reservation_status_date',inplace=True)
```

```python
plt.figure(figsize=(22,8))
plt.title('Average Daily Rate',size=20)
plt.plot(not_cancelled_df_adr['reservation_status_date'],not_cancelled_df_adr['adr'],label='not cancelled')
plt.plot(cancelled_df_adr['reservation_status_date'],cancelled_df_adr['adr'],label='cancelled')
plt.legend()
plt.show()
```
